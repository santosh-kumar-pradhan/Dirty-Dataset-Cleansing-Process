## Clean The Numerical Data

```python
# Remove special Character and Change dtype To Float
```

```python
df['Actual gross']=(
    df['Actual gross']
    .astype(str)
    .str.replace(r'[^0-9.]','',regex=True)
    .astype(float)
)
```

```python
df['Adjusted gross (in 2022 dollars)']=(
    df['Adjusted gross (in 2022 dollars)']
    .astype(str)   
    .str.replace(r'[^0-9.]','',regex=True)
    .astype(float)
    
)
```

```python
df['Average gross']=(
    df['Average gross']
    .astype(str)   
    .str.replace(r'[^0-9.]','',regex=True)
    .astype(float)
)
```

## Clean The Catagorical Data

### Unique Check

```python
df.Artist.unique()
```

```python
df['Tour title'].unique()
```

```python
df['Tour title'] = (
    df['Tour title']
    .astype(str)
    .str.replace(r'[^a-zA-Z\s&]', '', regex=True)  # keep letters, space, & symbol
    .str.strip()
)
```
